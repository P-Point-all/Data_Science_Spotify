## Data loading  
In the following project we will analyse a dataset with several data about some songs listened on Spotify.  
For each song we have the following variables :
- track_id
- track_name
- track_artist
- track_popularity : a grade between 0 and 100
- track_album_id : the id of the album of the track
- track_album_name
- track_album_release_date
- playlist_name
- playlist_genre
- danceability : float between 0 and 1 to quantify if the track is danceable, based on some criterion like tempo, rhythm stability, beat strength, and overall regularity
-  energy : float between 0 and 1 representing a perceptual measure of intencity and activity
-  key : estimated key of the track, 0=C, 1=C#, 2=D ... -1=no key detected
-  loudness : overall loudness in average of a track in decibels (dB), represents the quality of a sound
-  mode : modality of a track : major = 1, minor = 0
-  speechiness : float between 0 and 1 indicating the presence of spoken word in a track >2/3 : almost only spoken words, 1/3< <2/3 : contains both music and speech, <1/3 : music and non-speech-like tracks
-  acousticness : float between 0 and 1 describing if the song is accoustic
-  instrumentalness : float between 0 and 1 representing if the track is more vocal : close to 0 or intrumental : close to 1
-  liveness : float between 0 and 1 detecting the presence or not of an audience. Above 0.8 the track is probably a live
-  valence : a city in France in the south of Lyon, but here a float between 0 and 1 mesuring the postiveness of a song (close to 1)
-  tempo : tempo in beats per minute (BPM)
-  duration_ms

```python
# pip install prince
# pip install datasets

import pandas as pd
import numpy as np
import prince
# import datasets

import matplotlib.pyplot as plt
import seaborn as sns
# from datasets import load_dataset
```

```python
spotify_dataframe = pd.read_csv("spotify_songs.csv")
```

```python
spotify_dataframe.head()
```

Cleaning data : removing useless columns

```python
spotify_dataframe.info()
```

Track_id, track_name, playlist_id, track_album_id are identifier variables that have no purpose for clustering or inference, thus we will delete them from our dataframe. 

```python
spotify_dataframe.drop('track_id', axis = 1, inplace = True)
spotify_dataframe.drop('track_name', axis = 1, inplace = True)
spotify_dataframe.drop('playlist_id', axis = 1, inplace = True)
spotify_dataframe.drop('track_album_id', axis = 1, inplace = True)
```

Changing release date to only keep the year. Indeed, the exact release date has no purpose for clustering or inference, but we will get more overall information only keeping the year and the month (we will be able to see if one genre of music was trendy for a given year for instance, or if some songs are more popular in summer).

Problem here : time data does not always have the same format : example 2012, 2012-07... We choose to keep everything and set month to 0 if we do not have the info.

```python
from datetime import datetime
#print(spotify_dataframe['track_album_release_date'])
years = []
months = []
#print(spotify_dataframe.size)
for i in range(spotify_dataframe.shape[0]):
    #print(spotify_dataframe['track_album_release_date'][i])
    #print(i)
    if((len(spotify_dataframe['track_album_release_date'][i]))==4):
        years.append(datetime.strptime(spotify_dataframe['track_album_release_date'][i], '%Y').year)
        months.append(0)
    elif((len(spotify_dataframe['track_album_release_date'][i]))==7):
        years.append(datetime.strptime(spotify_dataframe['track_album_release_date'][i], '%Y-%m').year)
        months.append(datetime.strptime(spotify_dataframe['track_album_release_date'][i], '%Y-%m').month)
    else:
        years.append(datetime.strptime(spotify_dataframe['track_album_release_date'][i], '%Y-%m-%d').year)
        months.append(datetime.strptime(spotify_dataframe['track_album_release_date'][i], '%Y-%m-%d').month)
spotify_dataframe['track_album_release_year'] = pd.Series(years)
spotify_dataframe['track_album_release_month'] = pd.Series(months)
```

Removing track album release date (which is now a bit redundant with the new columns for release year and month)

```python
spotify_dataframe.drop('track_album_release_date', axis = 1, inplace = True)
```

Factoring categorial variables

```python
spotify_dataframe['track_artist'] = spotify_dataframe['track_artist'].astype('category')
spotify_dataframe['track_album_name'] = spotify_dataframe['track_album_name'].astype('category')
spotify_dataframe['playlist_name'] = spotify_dataframe['playlist_name'].astype('category')
spotify_dataframe['playlist_genre'] = spotify_dataframe['playlist_genre'].astype('category')
spotify_dataframe['playlist_subgenre'] = spotify_dataframe['playlist_subgenre'].astype('category')
spotify_dataframe['mode'] = spotify_dataframe['mode'].astype('category')
spotify_dataframe['track_album_release_year'] = spotify_dataframe['track_album_release_year'].astype('category')
spotify_dataframe['track_album_release_month'] = spotify_dataframe['track_album_release_month'].astype('category')
```

Columns containing missing values

```python
print('Columns containing missing values are : ', list(spotify_dataframe.columns[spotify_dataframe.isnull().sum()!=0]))
```

Track artist

```python
print("Track artist - Rate of missing values: {:.2f}%".format(spotify_dataframe['track_artist'].isnull().sum()/spotify_dataframe.shape[0]*100))
```

The rate of missing values for track artist variable is really low, it makes no sense to remove the variable nor to set it to the most common track artist, we will thus keep it the way it is.

```python
print("Track album name - Rate of missing values: {:.2f}%".format(spotify_dataframe['track_album_name'].isnull().sum()/spotify_dataframe.shape[0]*100))
```

Same conclusion with track album name than with the track artist variable.

We have to be careful because variables key and track_album_release_month do have missing values, even though those are not
set to null but respectively -1 and 0. Thus we still have to check their rate of missing values.

```python
rate_mv_key = 0
for i in range(spotify_dataframe.shape[0]):
    if(spotify_dataframe['key'][i] == -1):
        rate_mv_key += (1/spotify_dataframe.shape[0])
print("Key - Rate of missing values : {:.2f}%".format(rate_mv_key*100))
```

We basically have no missing values for the key variable, no need to change anything.

```python
rate_mv_tarm = 0
for i in range(spotify_dataframe.shape[0]):
    if(spotify_dataframe['track_album_release_month'][i] == 0):
        rate_mv_tarm += (1/spotify_dataframe.shape[0])
print("Track album release month - Rate of missing values : {:.2f}%".format(rate_mv_tarm*100))
```

We have a lot of missing values for track album release month. We still have not decided what to do with this.
