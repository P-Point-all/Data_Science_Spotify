<h2 style="color:#2980b9; font-size:32px; font-family:'Trebuchet MS', sans-serif; margin-top:30px;">
    1. Nettoyage et exploration du dataset
</h2>

<p style="font-size:22px; margin-top:40px">Dans ce projet, nous allons analyser un jeu de données contenant plusieurs informations sur des chansons écoutées sur Spotify.  
Pour chaque chanson, nous avons les variables suivantes :</p>

<div style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    <ul>
        <li style="margin-top: 5px">track_id</li>
        <li style="margin-top: 5px">track_name</li>
        <li style="margin-top: 5px">track_artist</li>
        <li style="margin-top: 5px">track_popularity : une note entre 0 et 100</li>
        <li style="margin-top: 5px">track_album_id : l'identifiant de l'album de la musique</li>
        <li style="margin-top: 5px">track_album_name</li>
        <li style="margin-top: 5px">track_album_release_date</li>
        <li style="margin-top: 5px">playlist_name</li>
        <li style="margin-top: 5px">playlist_genre : genre de la playlist dans laquelle est la musique</li>
        <li style="margin-top: 5px">danceability : flottant entre 0 et 1 pour quantifier si la musique est dansable, sur la base de critères tels que le tempo, la stabilité du rythme, la force des battements et la régularité générale</li>
        <li style="margin-top: 5px">energy : flottant entre 0 et 1 représentant une mesure perceptuelle de l'intensité et de l'activité</li>
        <li style="margin-top: 5px">key : estimation de la tonalité du morceau, 0=C, 1=C#, 2=D ... -1=aucune tonalité détectée</li>
        <li style="margin-top: 5px">loudness :intensité sonore en moyenne d'une musique en décibels (dB)</li>
        <li style="margin-top: 5px">mode : modalité d'une musique : majeur = 1, mineur = 0</li>
        <li style="margin-top: 5px">speechiness : flottant entre 0 et 1 indiquant la présence de paroles dans une piste >2/3 : presque uniquement des paroles, entre 1/3 et 2/3 : contient à la fois de la musique et de la parole, inférieur à 1/3 : pistes musicales et non parlantes</li>
        <li style="margin-top: 5px">acousticness : flottant entre 0 et 1 décrivant si la chanson est acoustique</li>
        <li style="margin-top: 5px">instrumentalness : flottant entre 0 et 1 représentant si la piste est plus vocale : proche de 0 ou instrumentale : proche de 1</li>
        <li style="margin-top: 5px">liveness : flottant entre 0 et 1 détectant la présence ou non d'un public. Au dessus de 0.8 la piste est probablement un live d'un concert</li>
        <li style="margin-top: 5px">valence : flotteur entre 0 et 1 mesurant la positivité d'une chanson (proche de 1)</li>
        <li style="margin-top: 5px">tempo : tempo en battements par minute (BPM)</li>
        <li style="margin-top: 5px">duration_ms: durée de la musique en millisecondes</li> 
    </ul>
</div>

```r
path <- "spotify_songs.csv"
song <- read.csv(path, header = TRUE)
# Premières lignes du jeu de données
head(song)
```

<div id="dupl"></div>
<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.1 Gestion des duplicatas
</h3>

On va tout d'abord regarder si on a des duplicatas dans notre dataset en utilisant le track_id.

```r
sum(duplicated(song$track_id))
#song<-song[!duplicated(song$track_id),]
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On voit qu'on a beaucoup de duplicatas. En cherchant dans le dataset, on a découvert que ces duplicatas sont enfait dans différentes playlists qui n'ont pas le même genre. On en déduit que l'extraction des données a été faites via les playlists et c'est donc pour cela qu'on a des duplicatas. Nous allons garder qu'une seule occurence de chaque musique. Pour choisir laquelle, nous allons garder une occurence dont le genre (puis le sous-genre) est le mode des différents genres de la musique (différents genres des différentes occurences de la musique). Exemple: Dancing Queen apparait 3 fois, 2 fois avec pop et une fois avec latin, on va garder une occurence de pop.
On en profite également pour stocker dans une nouvelle variable le nombre de playlist où apparait la musique. Cependant, une musique présente dans de nombreuses playlists implique que cette musique doit être plutôt écouté et pour ne pas perdre non plus toutes l'information, on va garder pour chaque musique son nombre d'apparitions dans le dataset.
<br/><br/>
Ce nettoyage particulier est réalisé notamment pour les analyses factorielles afin de ne pas trop les biaiser avec une musique apparaissant plus que d'autres. Pour certaines méthodes, notamment la NMF, on va utiliser le dataset complet pour obtenir de meilleurs résultats.

```r
song_complete=data.table::copy(song) #on garde une version du dataset complet 

song <- song %>%
  group_by(track_id) %>%
  add_count(name = "playlist_appearance_count") %>%
  ungroup() %>%

  # Ensuite ton pipeline :
  group_by(track_id, playlist_genre) %>%
  add_count(name = "genre_count") %>%
  group_by(track_id) %>%
  arrange(desc(genre_count), .by_group = TRUE) %>%
  slice(1) %>%
  ungroup() %>%
  dplyr::select(-genre_count) %>%

  group_by(track_id, playlist_subgenre) %>%
  add_count(name = "subgenre_count") %>%
  group_by(track_id) %>%
  arrange(desc(subgenre_count), .by_group = TRUE) %>%
  slice(1) %>%
  ungroup() %>%
  dplyr::select(-subgenre_count)
sum(duplicated(song$track_id)) #vérification que l'on a bien retiré tous les duplicatas
```

<div id="trans"></div>
<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2 Transformation sur les variables
</h3>
<h4 style="color:#2980b9; font-size:20px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2.1 Transformation de la date
</h4>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On va décomposer la variable de date en jour mois année puis uniquement récupérer le mois et l'année. Si le mois n'est pas présent, il est mit par défaut à 1, on ne va pas s'intéresser trop à cette variable donc cela ne pose pas de problème

```r
formats <- c("Ymd", "Ym", "Y")

# Parse les dates en tenant compte des différents formats
song$parsed_date <- parse_date_time(song$track_album_release_date, orders = formats)

# Extraire mois et année
song$month <- month(song$parsed_date, label = FALSE) # label=TRUE pour Jan, Feb...
song$year <- year(song$parsed_date)

# Supprimer l'ancienne colonne
song <- subset(song, select = -track_album_release_date)

# (optionnel) supprimer la colonne parsed_date si plus utile
song <- subset(song, select = -parsed_date)
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    Nous allons maintenant supprimer les variables inutiles à l'analyse comme le `track_id` (un identifiant différent par musique) qui ne sera d'aucune utilité pour notre analyse car l'information est déjà contenue dans la variable `track_name` que nous n'utiliserons que pour labelliser.
De même, on va supprimer `track_album_id`,`playlist_id`.

<h4 style="color:#2980b9; font-size:20px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2.2 Changement d'unité sur la durée
</h4>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On va également modifier la variable de durée en la passant en secondes au lieu de millisecondes pour avoir des valeurs plus parlantes.

```r
song<-subset(song,select=-track_id)
song<-subset(song,select=-track_album_id)
song<-subset(song,select=-playlist_id)

colnames(song)[which(colnames(song)=="duration_ms")]<-"duration_s"
song[,"duration_s"]<-song[,"duration_s"]/1000

head(song)
```

<h4 style="color:#2980b9; font-size:20px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2.3 Transformation des variables qualitatives en facteurs
</h4>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On modifie les variables qualitatives en facteurs. On en utilisera certaines pour les analyses, d'autres ne seront utilisés que pour labelliser.

```r
song$playlist_genre<-as.factor(song$playlist_genre)
song$playlist_subgenre<-as.factor(song$playlist_subgenre)
song$mode<-as.factor(song$mode)
song$track_artist<-as.factor(song$track_artist)
song$track_album_name<-as.factor(song$track_album_name)
song$playlist_name<-as.factor(song$playlist_name)
song$key<-as.factor(song$key)
head(song)
```

```r
colSums(is.na(song))
```

<h4 style="color:#2980b9; font-size:20px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2.4 Gestion des valeurs manquantes
</h4>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    Comme les NA pour le mois on été géré en les remplacant par une valeur par défaut, on a que quelques NA pour les variables `track_artist` et `track_album_name`. On va les retirer pour éviter des problèmes d'éxécution lors des différentes analyses. On ne perd que très peu d'information (4 valeurs sur tous les datasets).

```r
song<-song %>% drop_na(track_artist)
song_complete<-song_complete %>% drop_na(track_artist)
```

```r
colSums(is.na(song))
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On voit qu'on a bien supprimé tous les individus sans track_artist et track_album_name.

```r
nb_mv_key <- 0

for (i in 1:nrow(song)) {
  if (song$key[i] == -1) {
    nb_mv_key <- nb_mv_key + 1
  }
}

cat("Key - Number of missing values :", nb_mv_key, "\n")
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">Il n'y a aucune valeur manquante pour la clé, contrairement à ce qu'on aurait pu supposer avec le descriptif du jeu de données. Cela fait un problème de moins à régler.

<h4 style="color:#2980b9; font-size:20px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
1.2.5 Gestion des valeurs aberrantes
</h4>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">On va supprimer les musiques qui durent moins de 30 secondes car elles sont trop courtes pour qu'on puisse vraiment les juger.

```r
song<-song[!song$duration_s<30,]
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">On va créer une variable period qui représente la décennie de sortie de la musique.

```r
song <- song %>%
  mutate(
    period = case_when(
      is.na(year)              ~ "2010s",
      year > 2010              ~ "2010s",
      year > 2000              ~ "2000s",
      year > 1990              ~ "1990s",
      year > 1980              ~ "1980s",
      year > 1970              ~ "1970s",
      year > 1960              ~ "1960s",
      TRUE                     ~ "1950s"
    ))
song$month<-as.factor(song$month)
song$year<-as.factor(song$year)
song$period <- as.factor(song$period)
```

<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
2.1 Correlation plot
</h3>

```r
histo_genre <- ggplot(song,aes(x=playlist_genre))+
  geom_histogram(stat="count")+
  labs(title = "Distribution des genres des musiques")

histo_year <- ggplot(song,aes(x=year))+
    geom_histogram(stat="count")+
    labs(title = "Distribution des années de sortie des musiques")+
    scale_x_discrete(breaks = levels(factor(song$year))[seq(1, length(levels(factor(song$year))), by = 10)])

histo_month <- ggplot(song,aes(x=month))+
    geom_histogram(stat="count",fill='blue')+
    labs(title = "Distribution des mois de sortie des musiques")

histo_popularity <- ggplot(song,aes(x=track_popularity))+
    geom_histogram(stat="count")+
    labs(title = "Distribution des popularités des musiques")

histo_danceability <- ggplot(song,aes(x=danceability))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de la danceabilité des musiques")

histo_energy <- ggplot(song,aes(x=energy))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de l'energie des musiques")

histo_key <- ggplot(song,aes(x=key))+
    geom_histogram(stat='count')+
    labs(title = "Distribution de la clé des musiques")

histo_loudness <- ggplot(song,aes(x=loudness))+
    geom_histogram(bins=50)+
  labs(title = "Distribution de la force sonore des musiques")

histo_mode  <- ggplot(song,aes(x=mode))+
    geom_histogram(bins=50,stat='count')+
    labs(title = "Distribution de du mode des musiques")

histo_speechiness <- ggplot(song,aes(x=speechiness))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de la vocabilité des musiques")

histo_instrumentalness <- ggplot(song,aes(x=instrumentalness))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de l'intrumentabilité des musiques")

histo_liveness <- ggplot(song,aes(x=liveness))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de la liveness des musiques")

histo_valence <- ggplot(song,aes(x=valence))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de la valence des musiques")

histo_tempo <- ggplot(song,aes(x=tempo))+
    geom_histogram(bins=50)+
    labs(title = "Distribution du tempo des musiques")

histo_duration_s <- ggplot(song,aes(x=duration_s))+
    geom_histogram(bins=50)+
    labs(title = "Distribution de la durée des musiques")

histo_nbplaylist  <- ggplot(song,aes(x=playlist_appearance_count))+
    geom_histogram()+
    labs(title = "Distribution du nombre d'apparition dans les playlists")
histo_period <- ggplot(song,aes(x=period))+
    geom_histogram(stat='count')+
    labs(title ="Distribution des décennies des musiques")
```

```r
options(repr.plot.width=20, repr.plot.height=25)
grid.arrange(histo_genre,histo_month,histo_year, histo_popularity,histo_danceability,histo_energy,histo_key,histo_loudness,histo_mode,histo_speechiness, histo_instrumentalness, histo_liveness,histo_valence,histo_tempo,histo_duration_s,histo_nbplaylist,histo_period, ncol=3)
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">On observe qu'on a beaucoup plus de musiques récentes notamment de la dernière décennie (2010-2020), ce qui ce comprend car c'est à ce moment qu'on émergé les plateformes de streaming comme Spotify et que le marché de la musique a explosé.
<br/><br/>
On a beaucoup de musique très peu populaire. La majorité des gens vont écouter des artistes connus mais comme il est assez aisé (comparé à avant) de publier une musique sur les plateformes de streaming, il y a beacoup d'artistes niches et donc de musiques très peu voir pas écouté donc avec une très faible popularité.
<br/><br/>
Assez surprenament, on a beaucoup moins de musiques avec la tonalité (D♯/E♭) que les autres tonalité, aucune explication pour cela.
<br/><br/>
On a dans notre jeu de données des musiques avec assez peu d'élocution (speechiness) et d'instrumentalité. On en déduit que ces mesures sont assez dures vis à vis des musiques. Il semble assez compliqué d'avoir une speechinness ou une instrumentalness assez élevé. On a une certaine contradiction avec la descprition du dataset qui nous a été donné car ces deux mesures devrait ce complémenter or on voit bien sur le graphe que ce n'est pas le cas. On peut avoir des musiques avec une speechiness et instrumentalness très faible ce qui correspond à une musique où il y a peu d'instrument/fond sonore et peu de paroles donc enfait des musiques vides.

3.3 CAH
</h3>

```r
library(cluster)
# Subsample the projected data by taking every 5th row
projected_sampled <- projected[seq(1, nrow(projected), by = 5), ]

# Print the dimensions of the full and subsampled data
cat("Taille du jeu de données complet dans l'espace d'ACP:", dim(projected), "\n")
cat("Taille du sous-échantillon dans l'espace d'ACP:", dim(projected_sampled), "\n")
```

```r
# Perform agglomerative clustering with ward linkage
ac <- hclust(dist(projected_sampled), method = "ward.D2")

# Extract the distances at which clusters are merged
distances <- ac$height

# Number of sizes to plot
n_sizes <- 20

# Prepare data for plotting
x <- seq(n_sizes+1, 2, by = -1)
y <- tail(distances, n_sizes)

# Create a data frame for plotting
plot_data <- data.frame(x = x, y = y)

# Plot the distances
ggplot(plot_data, aes(x = x, y = y)) +
  geom_point() +
  labs(x = "Nombre de classes", y = "Mesure d'aggrégation", title = "Choix du nombre de classes") +
  theme_minimal()
```

```r
d <- dist(projected_sampled)  # Euclidean distance

# Hierarchical clustering with Ward's method
hc <- hclust(d, method = "ward.D2")  # ward.D2 is equivalent to 'ward' in Python

# Plot dendrogram
plot(hc, labels = FALSE, main = "Dendrogramme avec lien de Ward")

# Cutting the dendrogram to get K = 6 clusters
K <- 6
clusters_ac <- cutree(hc, k = K)

# Optional: Highlight clusters in the dendrogram
# rect.hclust(hc, k = K, border = "black")

# Calculate height for horizontal cut (optional, for visualization)
heights <- sort(hc$height, decreasing = TRUE)
max_d <- 0.5 * (heights[K] + heights[K-1])
abline(h = max_d, col = "red", lty = 2)
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">On a choisi le lien de Ward car il a tendance à construire des classes de taille similaire avec des centres de gravité proches et car il favorise les classes sphériques.
<br/><br/>
On voit que rien n'indique une séparation optimale en 6 classes pour CAH sur le dendrogramme, or il y a bien 6 genres dans notre jeu de données.

<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
3.4 Matrices de confusion sur les différentes méthodes de clustering
</h3>

```r
correspondence_score <- function(classif1, classif2) {
  # Création de la matrice de confusion
  cm <- table(classif1, classif2)
  
  # Calcul du total
  total <- sum(cm)
  
  # Initialisation de la somme
  sum_matches <- 0
  
  # Boucle pour trouver les correspondances maximales
  for (i in 1:6) {
    # Trouver l'indice du maximum
    max_idx <- which(cm == max(cm), arr.ind = TRUE)[1, ]
    
    # Ajouter la valeur maximale à la somme
    sum_matches <- sum_matches + cm[max_idx[1], max_idx[2]]
    
    # Supprimer la ligne et la colonne
    cm <- cm[-max_idx[1], -max_idx[2], drop = FALSE]
  }
  
  # Retourner le score
  return(sum_matches/total)
}

# Pour visualiser une matrice de confusion (équivalent à ConfusionMatrixDisplay)
plot_confusion_matrix <- function(classif1, classif2) {
  cm <- table(classif1, classif2)
  
  # Création du graphique avec heatmap
  heatmap(as.matrix(cm), 
          Rowv = NA, 
          Colv = NA, 
          main = "Matrice de confusion", 
          xlab = "Prédiction", 
          ylab = "Référence",
          cexRow = 0.8, 
          cexCol = 0.8)
}
```

```r
playlist_genre_num <- song$playlist_genre

# Pour KMeans
cm_kmeans <- table(playlist_genre_num, clusters_pca)

# Alternative plus proche de ConfusionMatrixDisplay avec le package caret
library(corrplot)
corrplot(as.matrix(cm_kmeans), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - KMeans",
         xlab = "Sur les données (KMeans)", ylab = "Playlist genres")

# Affichage du score pour KMeans
cat(paste("Score de correspondance pour KMeans :", 
          round(100 * correspondence_score(playlist_genre_num, clusters_pca), 2), "%\n"))

# Pour CAH
cm_cah <- table(playlist_genre_num[seq(1, length(playlist_genre_num), 5)], clusters_ac)
corrplot(as.matrix(cm_cah), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - CAH",
         xlab = "Sur les données (CAH)", ylab = "Playlist genres")

# Affichage du score pour CAH
cat(paste("Score de correspondance pour CAH :", 
          round(100 * correspondence_score(playlist_genre_num[seq(1, length(playlist_genre_num), 5)], clusters_ac), 2), "%\n"))

# Pour GMM
cm_gmm <- table(playlist_genre_num, clusters_gmm)
corrplot(as.matrix(cm_gmm), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - GMM",
         xlab = "Sur les données (GMM)", ylab = "Playlist genres")

# Affichage du score pour GMM
cat(paste("Score de correspondance pour GMM :", 
          round(100 * correspondence_score(playlist_genre_num, clusters_gmm), 2), "%\n"))
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    Les différentes méthodes de clustering ne semblent pas efficaces pour retrouver les genres de `playlist_genre`. Même en imposant 6 classes qui est égal au nombre de genres différents, les matrices de confusion et les scores de correspondance calculés montrent une incapacité à relier les deux. Cela indique que les genres ne correspondent pas à aux clusters "naturels" des musiques.

<h2 style="color:#2980b9; font-size:32px; font-family:'Trebuchet MS', sans-serif; margin-top:30px;">
    4. Interprétation des clusters avec une MCA
</h2>

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">Si les clusters ne correspondent pas au genre, on peut se poser la question de ce qu'ils regroupent. On va donc réaliser une MCA en discrétisant certaines des variables quantitatives pour essayer de comprendre ce que représente les clusters.
<br/><br/>
On va discrétiser les variables acousticness, energy, instrumentalness et valence qui sont des variables plutôt bien représentés dans le premier plan factoriel de l'ACP, on s'attend donc, comme les clusterings ont été réalisé sur les coordonnées d'ACP à ce que les catégories soit plutôt bien distinctes.
<br/><br/>
On va faire une MCA par méthode de clustering pour éviter que la MCA ne cherche trop à faire des correspondaces entres les clustering au lieu de chercher des correspondances avec les modalités des variables qualitatives.

```r
song_MCA <- song %>%
  dplyr::select(playlist_genre)

# Ajout des variables discrétisées
song_MCA <- song_MCA %>%
  mutate(
    acousticness_class = cut(song$acousticness, 
                             breaks = c(0, 0.33, 0.66, 1), 
                             labels = c("faible", "moyen", "élevé"),
                             include.lowest = TRUE),
    
    energy_class = cut(song$energy, 
                       breaks = c(0, 0.33, 0.66, 1), 
                       labels = c("faible", "moyen", "élevé"),
                       include.lowest = TRUE),
    
    instrumentalness_class = cut(song$instrumentalness, 
                                 breaks = 3, 
                                 labels = c("faible", "moyen", "élevé"),
                                 include.lowest = TRUE),
    
    valence_class = cut(song$valence, 
                        breaks = 3, 
                        labels = c("faible", "moyen", "élevé"),
                        include.lowest = TRUE),
    
    clusters_kmeans = as.factor(clusters_pca)
  )
```

```r
head(song_MCA)
```

```r
res.mca = MCA(song_MCA, ncp=11,graph=FALSE)
res.mca$eig
```

```r
g1<-plot(res.mca,invisible="ind")
g2<-plot(res.mca,axes=c(3,4),invisible="ind")
grid.arrange(g1,g2,ncol=2)
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    Au lieu de devoir regarder chaque plan un après l'autre, (et il y en a beaucoup) pour voir les liens entre les clusters et les modalités, on peut calculer les distances à chaque modalités. Ainsi, en utilisant les coordonées de MCA de chaque modalités, on va calculer leurs distances aux coordonnées de MCA des clusters et on résume cela dans une matrice. Plus la distance est faible, plus le lien entre le cluster et la modalité est élevé.

```r

# coords : matrice des coordonnées des modalités (par exemple après une MCA)
coords <- res.mca$var$coord  # res.mca est l'objet MCA issu de FactoMineR::MCA()

# Séparer les modalités des clusters
clusters_coords <- coords[19:24, ]
modalities_coords <- coords[-(19:24), ]

# Calcul des distances euclidiennes entre clusters et modalités
distances <- as.matrix(dist(rbind(clusters_coords, modalities_coords)))[rownames(clusters_coords), rownames(modalities_coords)]

# Affichage heatmap avec les valeurs
pheatmap(distances, 
         cluster_rows = FALSE, 
         cluster_cols = FALSE, 
         display_numbers = TRUE, 
         color = colorRampPalette(c("blue", "white", "red"))(100),
         main = "Distances entre clusters et modalités (MCA)")

```

```r
song_dec <- song

song_dec <- subset(song_dec, select = -c(
  track_artist,
    track_name,
  track_popularity,
  track_album_name,
  playlist_name,
    month,
  playlist_genre,
  playlist_subgenre,
  period,
  playlist_appearance_count,
   year
))

song_dec$year <- as.integer(song$year)

str(song_dec)
head(song_dec)
```

```r
song_dec <- song_dec[song_dec$year <= 2000, ]

song_dec$key <- as.integer(song_dec$key)
song_dec$mode <- as.integer(song_dec$mode)

str(song_dec)
```

```r
song_dec_wy <- subset(song_dec, select = -year)

song_dec_wy <- song_dec_wy[, c("danceability", "energy", "loudness", "speechiness", "acousticness", "instrumentalness",
                              "liveness", "valence", "key", "mode", "duration_s", "tempo")]

group_lengths <- c(
  physical = 8,
  music = 2,
  duration = 1,
  tempo = 1
)

mfa_result <- MFA(
  song_dec_wy,
  group = group_lengths,
  type = rep("s", length(group_lengths)), 
  name.group = names(group_lengths),
  ncp = 7,
  graph = FALSE
)
```

```r
mfa_result$eig
```

On choisit de conserver 7 dimensions pour expliquer 80% de la variance.

<div id="62a"></div>
<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
6.2 Régressions linéaires
</h3>

```r


X <- as.data.frame(mfa_result$ind$coord)

y <- song_dec$year

train_indices <- createDataPartition(y, p = 0.67, list = FALSE)
X_train <- X[train_indices, ]
X_test  <- X[-train_indices, ]
y_train <- y[train_indices]
y_test  <- y[-train_indices]

model <- lm(y_train ~ ., data = X_train)

y_pred <- predict(model, newdata = X_test)

mae <- mean(abs(y_pred - y_test))
cat("Mean Absolute Error:", round(mae, 2), "\n")

r2 <- cor(y_pred, y_test)^2
cat("Model score (R²):", round(r2, 2), "\n")
     
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">
    On a un $R^2$ très faible (0.15) en opérant une régression linéaire pour prédire l'année de sortie à partir des autres variables. Celle-ci est prédite correctement à environ 7 ans près.

```r
library(glmnet)
library(Metrics)
X_train_mat <- as.matrix(X_train)
X_test_mat <- as.matrix(X_test)

ridge_cv <- cv.glmnet(
  x = X_train_mat,
  y = y_train,
  alpha = 0,
  lambda = c(0.01, 0.1, 1.0, 10.0, 100, 1000),
  nfolds = 5
)

cat("Best lambda (alpha_):", ridge_cv$lambda.min, "\n")

y_pred <- predict(ridge_cv, newx = X_test_mat, s = ridge_cv$lambda.min)

mae_value <- mae(y_test, y_pred)
cat("Mean Absolute Error:", round(mae_value, 2), "\n")

r2 <- cor(y_test, y_pred)^2
cat("Model score (R²):", round(r2, 2), "\n")
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">Les résultats ne sont pas meilleurs en utilisant une régression Ridge où le paramètre de pénalisation L2 a été choisi par validation croisée.

```r
X_mat <- as.matrix(X)
X_test_mat <- as.matrix(X_test)
y <- song_dec$year

set.seed(0)
lasso_cv <- cv.glmnet(
  x = X_mat,
  y = y,
  alpha = 1,
  nfolds = 100
)

y_pred <- predict(lasso_cv, newx = X_test_mat, s = lasso_cv$lambda.min)

mae_val <- mae(y_test, y_pred)
cat("Mean Absolute Error:", round(mae_val, 2), "\n")

y_full_pred <- predict(lasso_cv, newx = X_mat, s = lasso_cv$lambda.min)
r2_full <- cor(y, y_full_pred)^2
cat("R² on full data:", round(r2_full, 2), "\n")
     
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">Les résultats sont légèrement meilleurs en opérant une pénalisation L1 (régression LASSO).

<div id="63a"></div>
<h3 style="color:#2980b9; font-size:28px; font-family:'Trebuchet MS', sans-serif; margin-top:20px;">
6.3 Régression avec un arbre binaire de décision
</h3>

```r
library(rpart)

train_control <- trainControl(method = "cv", number = 10)

tune_grid <- expand.grid(cp = 0)

best_mae <- Inf
best_depth <- NULL
best_model <- NULL

for (d in 1:15) {
  model <- rpart(
    y_train ~ .,
    data = cbind(X_train, y_train),
    method = "anova",
    control = rpart.control(maxdepth = d)
  )
  
  y_pred_cv <- predict(model, newdata = X_train)
  mae_val <- mean(abs(y_pred_cv - y_train))
  
  if (mae_val < best_mae) {
    best_mae <- mae_val
    best_depth <- d
    best_model <- model
  }
}

cat("La profondeur optimale pour l'arbre de décision est :", best_depth, "\n")

final_model <- rpart(
  y_train ~ .,
  data = cbind(X_train, y_train),
  method = "anova",
  control = rpart.control(maxdepth = best_depth)
)

y_pred <- predict(final_model, newdata = X_test)
mae_val <- mean(abs(y_pred - y_test))
r2_val <- cor(y_pred, y_test)^2

cat("Mean Absolute Error:", round(mae_val, 2), "\n")
cat("Model score (R²):", round(r2_val, 2), "\n")
```

<p style="font-size:16px; font-family:'Verdana', sans-serif; color:#2d3436; line-height:1.6; border-left: 5px solid #3498db; background-color: #ecf5fb; padding: 10px; margin: 15px 0;">Malgré l'utilisation de validation croisée pour déterminer la profondeur optimale de notre arbre de régression, les résultats de prédiction sont encore moins bons ici que pour les régressions linéaires.

```r
correspondence_score <- function(classif1, classif2) {
  # Création de la matrice de confusion
  cm <- table(classif1, classif2)
  
  # Calcul du total
  total <- sum(cm)
  
  # Initialisation de la somme
  sum_matches <- 0
  
  # Boucle pour trouver les correspondances maximales
  for (i in 1:6) {
    # Trouver l'indice du maximum
    max_idx <- which(cm == max(cm), arr.ind = TRUE)[1, ]
    
    # Ajouter la valeur maximale à la somme
    sum_matches <- sum_matches + cm[max_idx[1], max_idx[2]]
    
    # Supprimer la ligne et la colonne
    cm <- cm[-max_idx[1], -max_idx[2], drop = FALSE]
  }
  
  # Retourner le score
  return(sum_matches/total)
}

# Pour visualiser une matrice de confusion (équivalent à ConfusionMatrixDisplay)
plot_confusion_matrix <- function(classif1, classif2) {
  cm <- table(classif1, classif2)
  
  # Création du graphique avec heatmap
  heatmap(as.matrix(cm), 
          Rowv = NA, 
          Colv = NA, 
          main = "Matrice de confusion", 
          xlab = "Prédiction", 
          ylab = "Référence",
          cexRow = 0.8, 
          cexCol = 0.8)
}
```

```r
library(RcppML) #on utilise une autre librairie pour la nmf afin de l'accélérer
nmf_rcpp <- nmf(X_nmf, k = 16, tol = 1e-4, maxit = 200, verbose = FALSE)
W_rcpp <- nmf_rcpp$w
H_rcpp <- nmf_rcpp$h
```

```r
n_top_artists <- 5

plot_list <- list()

for (i in 1:nrow(H_rcpp)) {
  topic <- H_rcpp[i, ]
  
  top_artists_idx <- order(topic, decreasing = TRUE)[1:n_top_artists]

  top_artists_names <- list_artists[top_artists_idx]
  top_artists_scores <- topic[top_artists_idx]
  
  df_topic <- data.frame(
    artists = factor(top_artists_names, levels = rev(top_artists_names)), # rev pour inverser l'ordre
    scores = top_artists_scores
  )
  
  p <- ggplot(df_topic, aes(x = scores, y = artists)) +
    geom_col(fill = "steelblue", width = 0.7) +
    labs(title = paste("Topic", i), x = "", y = "") +
    theme_minimal() +
    theme(
      plot.title = element_text(size = 16, hjust = 0.5, face = "bold"),
      axis.text = element_text(size = 12),
      axis.text.x = element_text(size = 10),
      axis.text.y = element_text(size = 12),
      panel.grid.major.y = element_blank(),
      panel.grid.minor = element_blank(),
      axis.line.x = element_line(color = "black"),
      panel.border = element_blank()
    )
  
  plot_list[[i]] <- p
}

grid_plot <- do.call(grid.arrange, c(plot_list, ncol = 4, nrow = 4))
```

    Comme dans le fichier Python les artistes qui contribuent le plus à caque topics sont généralement de genres musicaux plutôt proches. Cependant les topics sont différents car ils dépendent de la décompostion obtenue que n'est pas unique.  
On réalise à présent un algorithme de recommandation.

```r
artists_similarities <- cosine((H_rcpp))
```

```r
recommendation <- function(artist_name, similarities_matrix=artists_similarities, artists_list=list_artists, n_rec = 10) {
  artist_idx <- which(tolower(artists_list) == tolower(artist_name))
  
  if (length(artist_idx) == 0) {
    cat("Artiste non trouvé:", artist_name, "\n")
    return(invisible(NULL))
  }
  
  artists_sim <- similarities_matrix[artist_idx, ]
  suggestions_idx <- order(artists_sim, decreasing = TRUE)[2:(n_rec + 1)]
  
  recommendations_df <- data.frame(
    Rang = 1:n_rec,
    Artiste = artists_list[suggestions_idx],
    Similarite = round(artists_sim[suggestions_idx], 4)
  )
  
  cat("Recommandations pour", artist_name, ":\n\n")
  print(recommendations_df, row.names = FALSE)
  cat("\n")
  
  return(invisible(NULL))
}
```

```r
recommendation("Imagine Dragons")
```
