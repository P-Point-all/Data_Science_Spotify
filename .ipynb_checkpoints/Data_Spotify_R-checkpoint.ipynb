### I.Exploration et nettoyage du dataset
### Data loading  
In the following project we will analyse a dataset with several data about some songs listened on Spotify.  
For each song we have the following variables :
- track_id
- track_name
- track_artist
- track_popularity : a grade between 0 and 100
- track_album_id : the id of the album of the track
- track_album_name
- track_album_release_date
- playlist_name
- playlist_genre
- danceability : float between 0 and 1 to quantify if the track is danceable, based on some criterion like tempo, rhythm stability, beat strength, and overall regularity
-  energy : float between 0 and 1 representing a perceptual measure of intensity and activity
-  key : estimated key of the track, 0=C, 1=C#, 2=D ... -1=no key detected
-  loudness : overall loudness in average of a track in decibels (dB), represents the quality of a sound
-  mode : modality of a track : major = 1, minor = 0
-  speechiness : float between 0 and 1 indicating the presence of spoken word in a track >2/3 : almost only spoken words, 1/3< <2/3 : contains both music and speech, <1/3 : music and non-speech-like tracks
-  acousticness : float between 0 and 1 describing if the song is accoustic
-  instrumentalness : float between 0 and 1 representing if the track is more vocal : close to 0 or intrumental : close to 1
-  liveness : float between 0 and 1 detecting the presence or not of an audience. Above 0.8 the track is probably a live
-  valence :  a float between 0 and 1 mesuring the positiveness of a song (close to 1)
-  tempo : tempo in beats per minute (BPM)
-  duration_ms

```r
path <- "spotify_songs.csv"
song <- read.csv(path, header = TRUE)
# Premières lignes du jeu de données
head(song)
# Vérification du contenu
summary(song)
```

On va tout d'abord regarder si on a des duplicatas dans notre dataset en utilisant le track_id.

```r
sum(duplicated(song$track_id))
```

On voit qu'on a beaucoup de duplicatas. En cherchant dans le dataset, on a découvert que ces duplicatas sont enfait dans différentes playlists qui n'ont pas le même genre. On en déduit que l'extraction des données a été faites via les playlists et c'est donc pour cela qu'on a des duplicatas. Néanmoins, nous décidons de les supprimmer et de garder uniquement la première occurence dans le jeu de données et on gardera la playlist et le genre correspondant.

```r
library(dplyr)

# Keep a complete version of the dataset
song_complete <- song

# 1. Count how many times each track_id appears
song <- song %>%
  add_count(track_id, name = "playlist_appearance_count")

# 2. Keep the most frequent genre for each track_id
genre_counts <- song %>%
  group_by(track_id, playlist_genre) %>%
  summarise(genre_count = n(), .groups = 'drop') %>%
  group_by(track_id) %>%
  slice(which.max(genre_count))

# 3. Merge with song to keep only rows corresponding to the dominant genre
song <- song %>%
  inner_join(genre_counts %>% select(track_id, playlist_genre), by = c("track_id", "playlist_genre"))

# 4. Keep the most frequent subgenre for each track_id
subgenre_counts <- song %>%
  group_by(track_id, playlist_subgenre) %>%
  summarise(subgenre_count = n(), .groups = 'drop') %>%
  group_by(track_id) %>%
  slice(which.max(subgenre_count))

# 5. Merge with song to keep only rows corresponding to the dominant subgenre
song <- song %>%
  inner_join(subgenre_counts %>% select(track_id, playlist_subgenre), by = c("track_id", "playlist_subgenre"))

# 6. Check if there are any remaining duplicates of track_id
nb_duplicates <- sum(duplicated(song$track_id))
print(nb_duplicates)

# Keep a complete version of the dataset without duplicates
song_complete_without_duplicates <- song
```

# 3. Clustering et lien avec le genre

## 3.1. Clustering avec K-means dans l'espace d'ACP

```r
library(stats)
# Set the number of clusters
K <- 6

# Perform K-means clustering
kmeans_pca <- kmeans(projected, centers = K, nstart = 25, iter.max = 100, algorithm = "Hartigan-Wong")

# Get the cluster assignments
clusters_pca <- kmeans_pca$cluster

# Create a data frame for plotting
plot_data <- data.frame(x = projected[, 1], y = projected[, 2], cluster = as.factor(clusters_pca))

# Plot the clusters
ggplot(plot_data, aes(x = x, y = y, color = cluster)) +
  geom_point(size = 1) +
  scale_color_brewer(palette = "Set1") +
  ggtitle("Graphe d'individus - Couleur selon les classes des k-means") +
  theme_minimal() +
  theme(plot.title = element_text(hjust = 0.5)) +
  geom_hline(yintercept = seq(floor(min(projected[, 2])), ceiling(max(projected[, 2])), by = 1), color = "gray", linetype = "dashed", alpha = 0.5) +
  geom_vline(xintercept = seq(floor(min(projected[, 1])), ceiling(max(projected[, 1])), by = 1), color = "gray", linetype = "dashed", alpha = 0.5)
```

Les classes obtenues avec k-means ne sont pas linéairement discernables dans le premier plan factoriel d'ACP.

## 3.2. Clustering avec CAH (Classification Ascendante Hiérarchique)

```r
# Subsample the projected data by taking every 5th row
projected_sampled <- projected[seq(1, nrow(projected), by = 5), ]

# Print the dimensions of the full and subsampled data
cat("Taille du jeu de données complet dans l'espace d'ACP:", dim(projected), "\n")
cat("Taille du sous-échantillon dans l'espace d'ACP:", dim(projected_sampled), "\n")
```

```r
# Perform agglomerative clustering with ward linkage
ac <- hclust(dist(projected_sampled), method = "ward.D2")

# Extract the distances at which clusters are merged
distances <- ac$height

# Number of sizes to plot
n_sizes <- 20

# Prepare data for plotting
x <- seq(n_sizes+1, 2, by = -1)
y <- tail(distances, n_sizes)

# Create a data frame for plotting
plot_data <- data.frame(x = x, y = y)

# Plot the distances
ggplot(plot_data, aes(x = x, y = y)) +
  geom_point() +
  labs(x = "Nombre de classes", y = "Mesure d'aggrégation", title = "Choix du nombre de classes") +
  theme_minimal()
```

```r
d <- dist(projected_sampled)  # Euclidean distance

# Hierarchical clustering with Ward's method
hc <- hclust(d, method = "ward.D2")  # ward.D2 is equivalent to 'ward' in Python

# Plot dendrogram
plot(hc, labels = FALSE, main = "Dendrogramme avec lien de Ward")

# Cutting the dendrogram to get K = 6 clusters
K <- 6
clusters_ac <- cutree(hc, k = K)

# Optional: Highlight clusters in the dendrogram
# rect.hclust(hc, k = K, border = "black")

# Calculate height for horizontal cut (optional, for visualization)
heights <- sort(hc$height, decreasing = TRUE)
max_d <- 0.5 * (heights[K] + heights[K-1])
abline(h = max_d, col = "red", lty = 2)
```

On a choisi le lien de Ward car il a tendance à construire des classes de taille similaire avec des centres de gravité proches et car il favorise les classes sphériques.

On voit que rien n'indique une séparation optimale en 6 classes pour CAH sur le dendrogramme, or il y a bien 6 genres dans notre jeu de données.

## 3.4. Matrices de confusion sur les différentes méthodes de clustering

```r
correspondence_score <- function(classif1, classif2) {
  # Création de la matrice de confusion
  cm <- table(classif1, classif2)
  
  # Calcul du total
  total <- sum(cm)
  
  # Initialisation de la somme
  sum_matches <- 0
  
  # Boucle pour trouver les correspondances maximales
  for (i in 1:6) {
    # Trouver l'indice du maximum
    max_idx <- which(cm == max(cm), arr.ind = TRUE)[1, ]
    
    # Ajouter la valeur maximale à la somme
    sum_matches <- sum_matches + cm[max_idx[1], max_idx[2]]
    
    # Supprimer la ligne et la colonne
    cm <- cm[-max_idx[1], -max_idx[2], drop = FALSE]
  }
  
  # Retourner le score
  return(sum_matches/total)
}

# Pour visualiser une matrice de confusion (équivalent à ConfusionMatrixDisplay)
plot_confusion_matrix <- function(classif1, classif2) {
  cm <- table(classif1, classif2)
  
  # Création du graphique avec heatmap
  heatmap(as.matrix(cm), 
          Rowv = NA, 
          Colv = NA, 
          main = "Matrice de confusion", 
          xlab = "Prédiction", 
          ylab = "Référence",
          cexRow = 0.8, 
          cexCol = 0.8)
}
```

```r
playlist_genre_num <- song$playlist_genre

# Pour KMeans
cm_kmeans <- table(playlist_genre_num, clusters_pca)

# Alternative plus proche de ConfusionMatrixDisplay avec le package caret
library(corrplot)
corrplot(as.matrix(cm_kmeans), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - KMeans",
         xlab = "Sur les données (KMeans)", ylab = "Playlist genres")

# Affichage du score pour KMeans
cat(paste("Score de correspondance pour KMeans :", 
          round(100 * correspondence_score(playlist_genre_num, clusters_pca), 2), "%\n"))

# Pour CAH
cm_cah <- table(playlist_genre_num[seq(1, length(playlist_genre_num), 5)], clusters_ac)
corrplot(as.matrix(cm_cah), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - CAH",
         xlab = "Sur les données (CAH)", ylab = "Playlist genres")

# Affichage du score pour CAH
cat(paste("Score de correspondance pour CAH :", 
          round(100 * correspondence_score(playlist_genre_num[seq(1, length(playlist_genre_num), 5)], clusters_ac), 2), "%\n"))

# Pour GMM
cm_gmm <- table(playlist_genre_num, clusters_gmm)
corrplot(as.matrix(cm_gmm), is.cor = FALSE, method = "color",
         addCoef.col = "black", tl.col = "black", tl.srt = 45,
         title = "Matrice de confusion - GMM",
         xlab = "Sur les données (GMM)", ylab = "Playlist genres")

# Affichage du score pour GMM
cat(paste("Score de correspondance pour GMM :", 
          round(100 * correspondence_score(playlist_genre_num, clusters_gmm), 2), "%\n"))
```

Les différentes méthodes de clustering ne semblent pas efficaces pour retrouver les genres de `playlist_genre`. Même en imposant 6 classes qui est égal au nombre de genres différents, les matrices de confusion et les scores de correspondance calculés montrent une incapacité à relier les deux. Cela indique que les genres ne correspondent pas à aux clusters "naturels" des musiques.

## 3.5. Et avec des méthodes supervisées?

### 3.5.1. Analyse Discriminante Linéaire

On va faire une LDA pour essayer de prédire le genre de la playlist dans laquelle se trouve la musique.

Pour faciliter la LDA, on va transformer les variables qualitatives en one-hot. On ne va garder que les propriétés physiques de la musique. On va supprimer aussi le sous-genre car si on ne connait pas le genre on ne connait pas non plus le sous-genre.
On va échelonner les données quantitatives comme pour l'ACP afin que les variables tempo et duration_s ne prennent pas trop d'importance par rapport aux autres variables qui ne prennent pas les mêmes échelles de valeurs.

```r
songLDA<-subset(song, select=-track_artist)
songLDA<-subset(songLDA, select=-track_album_name)
songLDA<-subset(songLDA, select=-playlist_name)
songLDA<-subset(songLDA, select=-playlist_subgenre)
songLDA<-subset(songLDA, select=-track_album_release_month)
songLDA<-subset(songLDA, select=-track_album_release_year)
songLDA<-subset(songLDA, select=-track_album_release_month_num)
songLDA<-subset(songLDA, select=-track_album_release_year_num)
songLDA<-subset(songLDA, select=-period)
songLDA<-subset(songLDA, select=-playlist_appearance_count)
songLDA<-subset(songLDA, select=-track_popularity)
summary(songLDA)
#songLDA<-subset(songLDA, select=-pop)
songLDA[,c(2:3,5,7:13)]<-scale(songLDA[,c(2:3,5,7:13)])
```

```r
head(songLDA)
```

```r
#install.packages('MASS')
```

```r
library(MASS)
set.seed(42)
sample<-sample(c(TRUE,FALSE),nrow(songLDA),replace=TRUE,prob=c(0.7,0.3))
train<-songLDA[sample,]
test<-songLDA[!sample,]
res.lda<-lda(playlist_genre~.,data=train)
```

```r
lda_pred<-predict(res.lda,test)
```

```r
library(caret)
conf_matrix <- confusionMatrix(lda_pred$class, factor(test$playlist_genre))
print(conf_matrix)
```

On obtient une précision de 46% sur la classification avec LDA dans l'espace de base, ce qui n'est pas très satisfaisant.

### 3.5.2. LDA dans l'espace d'ACP

```r
trainACP<-res.pca$ind$coord[sample,]
trainACP = as.data.frame(trainACP)
trainACP$playlist_genre = train$playlist_genre
testACP<-res.pca$ind$coord[!sample,]
testACP = as.data.frame(testACP)
res.lda<-lda(playlist_genre~.,data=trainACP)
```

```r
lda_pred<-predict(res.lda,testACP)
```

```r
conf_matrix <- confusionMatrix(lda_pred$class, factor(test$playlist_genre))
print(conf_matrix)
```

Les résultats sont encore moins bons pour LDA dans l'espace d'ACP ce qui suggère de rester dans l'espace de base pour la classification.

## CA

#### Nous transformons la variable quantitative `track_popularity` en variable qualitative, afin de l'utiliser dans une CA sur elle et la variable `playlist_genre `. Pour vérifier l'indépendance,on utilise un test du $\chi^2$.

```r
library(gplots)
library(graphics)
library(ggpubr)
```

```r
song$track_popularity_quali <- as.factor(floor((song$track_popularity)/10))
```

```r
table_genre_popu<-table(song$playlist_genre,song$track_popularity_quali)
```

```r
options(repr.plot.width=8,repr.plot.height=8)
balloonplot(t(table_genre_popu))
mosaicplot(table_genre_popu, shade = TRUE, las=2, main = "Genre and popularity")
```

> On observe qu'il y a très peu de sons populaires et les dépendances entres facteurs ne sont pas flagrantes. Nous allons mainteneant réalisé un test du $\chi^2$ pour voir si il y a indépendance et si la CA est utile ou non.

```r
chisq.test(table_genre_popu)
```

> La p-valeur est inférieur au niveau de confiance 5%, on rejette donc l'hypothèse d'indépendance à se niveau de confiance. On va donc pouvoir faire une CA pour trouver ces dépendances.

```r
res.ca = CA(table_genre_popu, graph=FALSE)
```

```r
res.ca$eig
```

> On voit qu'avec 2 dimensions on conserve plus de 90% de la variance ce qui est très satisfaisant. On va maintenant plot les facteurs dans le premier plan de CA.

```r
options(repr.plot.width=16,repr.plot.height=10)
fviz_ca_biplot(res.ca, repel=TRUE)
```

> On observe un lien entre les genre `edm`, `r&b` et les faibles popularités. On observe également un lien entre `rock`,`rap`,`latin` et les popularités moyennes. De mêmen `pop` a un lien avec les hautes popularité.
> On interprète cela comme le fait que le genre `pop` est un genre assez général, que tout le monde peut écouter assez souvent et aimer. Au contraire, les genres `edm` et `r&b` sont plus niches et ont une communauté plutôt restreinte. Les 3 derniers sont des genres avec des communautés plus importantes mais on ne peut pas forcément écouter ces musiques avec tout le monde et/ou dans toutes les situations.

## MCA

Pour étudier un impact potentiel du tempo dans certains intervalles, on va créer une variable `tempo_quali` de la même manière que la popularité dans la CA.

```r
song$tempo_quali <- NA 

for (i in 1:nrow(song)){
    if(song[i,]$tempo < 40){
        song[i,]$tempo_quali <- "Larghissimo"
    }
    else if(song[i,]$tempo <60 ){
        song[i,]$tempo_quali <- "Largo"
    }
    else if(song[i,]$tempo<80){
        song[i,]$tempo_quali <- "Adagio"
    }
    else if(song[i,]$tempo < 100){
        song[i,]$tempo_quali <- "Andante"
    }
    else if(song[i,]$tempo < 120){
        song[i,]$tempo_quali <- "Allegretto"
    }
    else if(song[i,]$tempo <140){
        song[i,]$tempo_quali <- "Vivace"
    }
    else if(song[i,]$tempo <188){
        song[i,]$tempo_quali <- "Presto"
    }
    else{
        song[i,]$tempo_quali<- "Prestissimo"
    }
}
song$tempo_quali <-as.factor(song$tempo_quali)
```

# 5. Etude du lien entre caractéristiques physiques d’un titre et période de sortie

## 5.1. Pre-processing avant la régression

```r
song_dec <- song

song_dec <- subset(song_dec, select = -c(
  track_artist,
  track_popularity,
  track_album_name,
  playlist_name,
  track_album_release_month,
    track_album_release_month_num,
  playlist_genre,
  playlist_subgenre,
  period,
  playlist_appearance_count,
    track_album_release_year
))

song_dec$track_album_release_year <- as.integer(song_dec$track_album_release_year)
song_dec <- subset(song_dec, select = -c(track_album_release_year_num))

str(song_dec)
head(song_dec)
```

Un problème que nous avons est que notre jeu de données n'est pas uniforme en terme d'années représentées. En effet comme vu précédemment les années >= 2010 sont sur-représentées par rapport aux autres. Cela pose problème car le régresseur que l'on va créer risquerait de toujours proposer des années entre 2000 et 2020. Il aurait ainsi de bonnes performances théoriques mais ne serait pas satisfaisant car ne prenant pas vraiment en compte les années 50 à 2000 qui nous intéressent. On va donc se restreindre aux musiques dont l'année de sortie est avant 2000.

```r
song_dec <- song_dec[song_dec$track_album_release_year <= 2000, ]

song_dec$key <- as.integer(song_dec$key)
song_dec$mode <- as.integer(song_dec$mode)

str(song_dec)
```

On opère une analyse multifactorielle sur le jeu de données privé de la variable année de sortie, avec les groupes suivants :
- caractéristiques physiques calculées par Spotify (danceability, speechiness, ...) dont les méthodes de calcul ne sont pas connues
- caractéristiques musicales (variables key et mode)
- durée
- tempo

```r
library(FactoMineR)
song_dec_wy <- subset(song_dec, select = -track_album_release_year)

song_dec_wy <- song_dec_wy[, c("danceability", "energy", "loudness", "speechiness", "acousticness", "instrumentalness",
                              "liveness", "valence", "key", "mode", "duration_s", "tempo")]

group_lengths <- c(
  physical = 8,
  music = 2,
  duration = 1,
  tempo = 1
)

mfa_result <- MFA(
  song_dec_wy,
  group = group_lengths,
  type = rep("s", length(group_lengths)), 
  name.group = names(group_lengths),
  ncp = 7,
  graph = FALSE
)
```

```r
mfa_result$eig
```

On choisit de conserver 7 dimensions pour expliquer 80% de la variance.
